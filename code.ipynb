# Exercise 2: Implement Read_Categories_Keywords()


```python
def Read_Categories_Keywords(file_name):
    with open (file_name, "r") as cat_file:
        LCategories = []
        LKeywords = []
        for line in cat_file:
            first_word = line.strip().split()[0][:-1]
            LCategories.append(first_word)
            
            keyword_set = set(line.strip().split(" ")[1:])
            LKeywords.append(keyword_set)
            
    return LCategories, LKeywords
```

# Exercise 3: Implement Document_Reader()

```python
def Document_Reader(filename):
    with open(filename, "r",  encoding='mac_roman') as file:
    
        text = file.read()
        
    text = text.lower()    
    LWords = []
    current_word = []
        
    valid = 'abcdefghijklmnñopqrstuvwxyz0123456789\''
        
    for char in text:
        if char in valid:
            current_word.append(char)
        else:                        
            if current_word:
                LWords.append(''.join(current_word))
                current_word = []
    if current_word:
        LWords.append(''.join(current_word))   
    return LWords
```

# Exercise 4: Implement Count_Frequencies()

```python
def Count_Frequencies(LWords, LCategories, LKeywords):
    
    Dfrequencies = {}
    
    for i, category in enumerate(LCategories):
        Dfrequencies[category] = {keyword: 0 for keyword in LKeywords[i]}
        
        for word in LWords:
            if word in Dfrequencies[category]:
                Dfrequencies[category][word] += 1
    return Dfrequencies 
```

For example, if DFrequencies is
• {'medical': {'physician': 0, 'infection': 0, 'patient': 0, 'care': 0},
'space': {'planet': 1, 'orbit': 0, 'nasa': 2},
'sports': {'medal': 0, 'record': 0, 'score': 0, 'team': 1, 'training': 0}
}

the output will be (3 occurrences of category 'space'
, so 75% and 1 occurrence of 'sports'
, so 25%):
• DVotes = {'medical': (0,0), 'space': (3,75), 'sports': (1,25)}

```python
def Classify_Doc(DFrequencies):
    #parameter initilisation
    DVotes = {}
    total_count = 0 
    max_percentage = 0
    max_category = None

    # calculating the counts
    for category in DFrequencies: #iterating over the categories in the dictionary
        category_count = 0

        for keyword in DFrequencies[category]: # iterating over the keywords in each category
            keyword_count = DFrequencies[category][keyword]
            category_count +=  keyword_count

        total_count += category_count
        DVotes[category] = (category_count,0)

    
    #calculating the percentages
    for category in DVotes:
        count = DVotes[category][0]

        if total_count > 0: #avoiding division by 0
            percentage = int(count/total_count * 100)
            DVotes[category] = (count, percentage)
        else:
            percentage = 0
            
        if percentage > max_percentage:
            max_percentage = percentage
            max_category = category

     
    #checking if any category has a percentage higher than 50
    if max_percentage > 50: 
        DocCategory = max_category
    else:
        DocCategory = "UNKNOWN" #setting the category to "UNKNOWN"
    
    return DVotes, DocCategory

DFrequencies = {'medical': {'physician': 0, 'infection': 0, 'patient': 0, 'care': 0},
'space': {'planet': 1, 'orbit': 0, 'nasa': 2},
'sports': {'medal': 0, 'record': 0, 'score': 0, 'team': 1, 'training': 0}
}
print(Classify_Doc(DFrequencies))
```

```python
'''
chat gpt's version

import os
from collections import Counter

def validate_file(prompt):
    """Prompt the user until a valid file is provided."""
    while True:
        file_name = input(prompt)
        if os.path.exists(file_name) and os.path.getsize(file_name) > 0:
            return file_name
        print("Invalid file. Please try again.")

def validate_folder(prompt):
    """Prompt the user until a valid folder is provided."""
    while True:
        folder_name = input(prompt)
        if os.path.exists(folder_name) and os.path.isdir(folder_name):
            return folder_name
        print("Invalid folder. Please try again.")

def process_document(file_path, LCategories, LKeywords):
    """Process a single txt document."""
    LWords = Document_Reader(file_path)
    DFrequencies = Count_Frequencies(LWords, LCategories, LKeywords)
    DVotes, DocCategory = Classify_Doc(DFrequencies)
    return DFrequencies, DVotes, DocCategory

def write_classification(file_output, file_name, DocCategory, DVotes, DFrequencies):
    """Write classification details for a single document."""
    if DocCategory != "UNKNOWN":
        file_output.write(f"{file_name}: {DocCategory} ({', '.join(f'{cat}:{perc}%' for cat, (count, perc) in DVotes.items())}): ")
        file_output.write(", ".join(f"{kw}:{count}" for kw, count in DFrequencies[DocCategory].items()) + "\n")
    else:
        file_output.write(f"{file_name}: UNKNOWN ({', '.join(f'{cat}:{perc}%' for cat, (count, perc) in DVotes.items())})\n")

def write_statistics(stats_file, total_files_number, category_counts, unknown_count, keyword_counts):
    """Write statistics to the output file."""
    stats_file.write("*** Statistics\n")
    stats_file.write(f"Total txt files: {total_files_number}\n")
    for category, count in category_counts.items():
        percentage = (count / total_files_number) * 100 if total_files_number > 0 else 0
        stats_file.write(f"{category}: {percentage:.2f}%\n")
    unknown_percentage = (unknown_count / total_files_number) * 100 if total_files_number > 0 else 0
    stats_file.write(f"UNKNOWN: {unknown_percentage:.2f}%\n")

    stats_file.write("*** Top 10 keywords\n")
    for keyword, count in keyword_counts.most_common(10):
        stats_file.write(f"{keyword}: {count}\n")

def main():
    # Step 1: Display welcome message
    DocClass_Welcome()

    # Step 2: Validate and load the categories and keywords file
    categories_file = validate_file("Enter the name of the file with categories and keywords: ")
    LCategories, LKeywords = Read_Categories_Keywords(categories_file)

    # Step 4: Validate and load the folder containing document files
    folder_name = validate_folder("Enter the name of the folder containing document files: ")

    # Step 5: Initialize variables for statistics
    total_files_number = 0
    unknown_count = 0
    category_counts = Counter()
    keyword_counts = Counter()

    classification_file = f"{folder_name}_DocClassification.txt"
    with open(classification_file, "w") as file_output:
        for file_name in os.listdir(folder_name):
            file_path = os.path.join(folder_name, file_name)

            if file_name.endswith('.txt'):
                # Process txt file
                DFrequencies, DVotes, DocCategory = process_document(file_path, LCategories, LKeywords)
                write_classification(file_output, file_name, DocCategory, DVotes, DFrequencies)

                # Update statistics
                total_files_number += 1
                if DocCategory != "UNKNOWN":
                    category_counts[DocCategory] += 1
                    for keyword, count in DFrequencies[DocCategory].items():
                        keyword_counts[keyword] += count
                else:
                    unknown_count += 1
            else:
                # Handle non-txt file
                file_output.write(f"Attention: the following non txt file has been detected: {file_name}\n")

    # Step 6: Write statistics
    statistics_file = f"{folder_name}_DCStatistics.txt"
    with open(statistics_file, "w") as stats_file:
        write_statistics(stats_file, total_files_number, category_counts, unknown_count, keyword_counts)

if __name__ == "__main__":
    try:
        main()
    except Exception as e:
        print(f"An error occurred: {e}")

'''
```
